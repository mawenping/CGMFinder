# CGMFinder — Quick Test

Run CGMFinder on the toy dataset (500 cells × 3000 genes).

```python
import os
import cgmfinder as cgmf

input_file = r"data/toy.feather"
output_dir = r"data/output"
os.makedirs(output_dir, exist_ok=True)
```

## Run with default parameters

```python
cgmf.run_analysis(
    input_file,
    output_dir,
    cpu_cores=5,
    ercc_correction=False
)
```

## Results

```python
import pickle

# List output files
for f in sorted(os.listdir(output_dir)):
    size = os.path.getsize(os.path.join(output_dir, f))
    print(f"{f:60s} {size/1024:.1f} KB")
```

```python
# Load identified modules
module_file = [f for f in os.listdir(output_dir) if f.endswith(".module.pickle")][0]
with open(os.path.join(output_dir, module_file), "rb") as f:
    modules = pickle.load(f)

print(f"Total modules: {len(modules)}")
for md, genes in sorted(modules.items(), key=lambda x: -len(x[1]))[:10]:
    print(f"  {md}: {len(genes)} genes")
```
